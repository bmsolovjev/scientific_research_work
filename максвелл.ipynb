# Уравнения Максвелла в среде

## Дифференциальная форма

$$
\boxed{
\begin{aligned}
&\nabla \cdot \mathbf{D} = \rho \\[8pt]
&\nabla \cdot \mathbf{B} = 0 \\[8pt]
&\nabla \times \mathbf{E} = -\frac{\partial \mathbf{B}}{\partial t} \\[8pt]
&\nabla \times \mathbf{H} = \mathbf{J} + \frac{\partial \mathbf{D}}{\partial t}
\end{aligned}
}
$$

## Где:

- $\mathbf{E}$ — напряжённость электрического поля (В/м)
- $\mathbf{D}$ — электрическая индукция (Кл/м²), $\mathbf{D} = \varepsilon \mathbf{E}$
- $\mathbf{H}$ — напряжённость магнитного поля (А/м)
- $\mathbf{B}$ — магнитная индукция (Тл), $\mathbf{B} = \mu \mathbf{H}$
- $\rho$ — объёмная плотность заряда (Кл/м³)
- $\mathbf{J}$ — плотность тока проводимости (А/м²)




## 1. Закон Гаусса для электричества

**Уравнение:**
$$ \nabla \cdot \mathbf{D} = \rho $$

**Операторы в трёхмерном виде:**
- $\nabla = \left( \dfrac{\partial}{\partial x}, \dfrac{\partial}{\partial y}, \dfrac{\partial}{\partial z} \right)$ — векторный оператор частных производных по трём координатам.
- $\nabla \cdot \mathbf{D} = \dfrac{\partial D_x}{\partial x} + \dfrac{\partial D_y}{\partial y} + \dfrac{\partial D_z}{\partial z}$ — **дивергенция** векторного поля $\mathbf{D}(x,y,z)$.

Дивергенция - дифференциальный оператор, отображающий векторное поле на скалярное. Он определяет, насколько расходится входящее и исходящее из малой окрестности данной точки поле, точнее, насколько расходятся входящий и исходящий потоки.

**Физический смысл:**  
В любой точке трёхмерного пространства дивергенция электрической индукции равна плотности свободного заряда.  
- $\nabla \cdot \mathbf{D} > 0$ — из точки «вытекает» больше поля, чем втекает → в точке есть положительный заряд (источник).  
- $\nabla \cdot \mathbf{D} < 0$ — поле «втекает» → отрицательный заряд (сток).  
- $\nabla \cdot \mathbf{D} = 0$ — заряда нет, поле не имеет источников в этой точке.

## 2. Закон Гаусса для магнетизма

**Уравнение:**
$$ \nabla \cdot \mathbf{B} = 0 $$

**Операторы в трёхмерном виде:**
- $\nabla \cdot \mathbf{B} = \dfrac{\partial B_x}{\partial x} + \dfrac{\partial B_y}{\partial y} + \dfrac{\partial B_z}{\partial z}$ — дивергенция магнитной индукции.

**Физический смысл:**  
В трёхмерном пространстве не существует точек, где дивергенция магнитного поля была бы отлична от нуля. Это означает, что **нет магнитных зарядов** (монополей).  
Линии магнитного поля всегда замкнуты (например, вокруг провода с током) или уходят в бесконечность, но не имеют начала и конца в конечных точках пространства. Поле не может ни «вытекать» из точки, ни «втекать» в неё.

---

## 3. Закон электромагнитной индукции Фарадея

**Уравнение:**
$$ \nabla \times \mathbf{E} = -\frac{\partial \mathbf{B}}{\partial t} $$

**Операторы в трёхмерном виде:**
- $\nabla \times \mathbf{E}$ — **ротор** (вихрь) электрического поля. В трёхмерных декартовых координатах вычисляется как определитель:
  $$ \nabla \times \mathbf{E} =
  \begin{vmatrix}
  \mathbf{i} & \mathbf{j} & \mathbf{k} \\
  \dfrac{\partial}{\partial x} & \dfrac{\partial}{\partial y} & \dfrac{\partial}{\partial z} \\
  E_x & E_y & E_z
  \end{vmatrix} =
  \left( \frac{\partial E_z}{\partial y} - \frac{\partial E_y}{\partial z},\;
         \frac{\partial E_x}{\partial z} - \frac{\partial E_z}{\partial x},\;
         \frac{\partial E_y}{\partial x} - \frac{\partial E_x}{\partial y} \right)
  $$
- $\dfrac{\partial \mathbf{B}}{\partial t} = \left( \dfrac{\partial B_x}{\partial t}, \dfrac{\partial B_y}{\partial t}, \dfrac{\partial B_z}{\partial t} \right)$ — частная производная каждой компоненты $\mathbf{B}$ по времени.

**Физический смысл:**  
Если в некоторой области трёхмерного пространства магнитное поле меняется во времени, то в этой области возникает **вихревое электрическое поле** — его ротор отличен от нуля.  
Знак «минус» (правило Ленца) означает, что возникающее электрическое поле стремится противодействовать изменению магнитного поля.  
Это ключевое уравнение для электромагнитной индукции, работы генераторов и трансформаторов.

---

## 4. Закон Ампера — Максвелла (с напряжённостью магнитного поля $\mathbf{H}$)

**Уравнение:**
$$ \nabla \times \mathbf{H} = \mathbf{J} + \frac{\partial \mathbf{D}}{\partial t} $$

**Операторы в трёхмерном виде:**
- $\nabla \times \mathbf{H}$ — ротор напряжённости магнитного поля, вычисляется аналогично ротору $\mathbf{E}$ через компоненты $H_x, H_y, H_z$.
- $\mathbf{J} = (J_x, J_y, J_z)$ — вектор плотности тока проводимости (заряды, движущиеся в пространстве).
- $\dfrac{\partial \mathbf{D}}{\partial t} = \left( \dfrac{\partial D_x}{\partial t}, \dfrac{\partial D_y}{\partial t}, \dfrac{\partial D_z}{\partial t} \right)$ — **плотность тока смещения**.

**Физический смысл:**  
В трёхмерном пространстве магнитное поле создаётся двумя причинами:
1. **Токами проводимости** $\mathbf{J}$ — движением свободных зарядов (например, электронов в проводе).
2. **Токами смещения** $\partial \mathbf{D}/\partial t$ — переменным во времени электрическим полем (даже в вакууме, где $\mathbf{J}=0$).

Ток смещения «замыкает» цепь переменного тока в конденсаторе и, главное, обеспечивает существование электромагнитных волн в трёхмерном пространстве: меняющееся электрическое поле порождает магнитное, а меняющееся магнитное — электрическое, и этот процесс распространяется в пространстве.

### Подстановка плоской волны в уравнения Максвелла

Плоская, распространяющаюся вдоль вектора $\mathbf{k}$ волна:
$$
\mathbf{E}(\mathbf{r}, t) = \mathbf{E}_0 e^{i(\mathbf{k} \cdot \mathbf{r} - \omega t)}, \quad \mathbf{B}(\mathbf{r}, t) = \mathbf{B}_0 e^{i(\mathbf{k} \cdot \mathbf{r} - \omega t)}
$$
Где:
- $\mathbf{k}$ — волновой вектор ($k = \omega/c$).
- $\omega$ — циклическая частота.
- $\mathbf{E}_0, \mathbf{B}_0$ — постоянные векторы амплитуд.

#### 1. Свойства оператора набла ($\nabla$) для плоской волны
При подстановке такой экспоненциальной формы дифференцирование превращается в алгебраическое умножение:
- Набла заменяется на $i\mathbf{k}$: $\nabla \to i\mathbf{k}$
- Производная по времени заменяется на $-i\omega$: $\frac{\partial}{\partial t} \to -i\omega$

#### 2. Проверка уравнений (в вакууме, $\rho=0, \mathbf{J}=0$):

*   **Закон Гаусса ($\nabla \cdot \mathbf{E} = 0$):**
    $$i\mathbf{k} \cdot \mathbf{E} = 0 \implies \mathbf{k} \perp \mathbf{E}$$
    (Электрическое поле перпендикулярно направлению распространения).

*   **Закон Гаусса для магнетизма ($\nabla \cdot \mathbf{B} = 0$):**
    $$i\mathbf{k} \cdot \mathbf{B} = 0 \implies \mathbf{k} \perp \mathbf{B}$$
    (Магнитное поле также перпендикулярно направлению распространения).

*   **Закон Фарадея ($\nabla \times \mathbf{E} = -\partial \mathbf{B}/\partial t$):**
    $$i\mathbf{k} \times \mathbf{E} = -(-i\omega \mathbf{B}) \implies \mathbf{B} = \frac{1}{\omega} (\mathbf{k} \times \mathbf{E})$$
    Это уравнение связывает амплитуды и фазы полей. Оно показывает, что $\mathbf{B} \perp \mathbf{E}$ и $\mathbf{B} \perp \mathbf{k}$.

*   **Закон Ампера-Максвелла ($\nabla \times \mathbf{B} = \mu_0\epsilon_0 \partial \mathbf{E}/\partial t$):**
    $$i\mathbf{k} \times \mathbf{B} = \mu_0\epsilon_0 (-i\omega \mathbf{E}) = -\frac{i\omega}{c^2} \mathbf{E}$$

**Вывод:** Все три вектора $(\mathbf{E}, \mathbf{B}, \mathbf{k})$ взаимно перпендикулярны и образуют правую тройку векторов. Это и подтверждает, что плоская волна является решением.
