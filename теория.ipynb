# Теоретическая справка
## 1) Уравнения Максвелла в системе СГС (гауссова система)

В дифференциальной форме:

$$
\boxed{
\begin{aligned}
&\nabla \cdot \mathbf{D} = 4\pi\rho, \\
&\nabla \cdot \mathbf{B} = 0, \\
&\nabla \times \mathbf{E} = -\frac{1}{c}\frac{\partial \mathbf{B}}{\partial t}, \\
&\nabla \times \mathbf{H} = \frac{4\pi}{c}\mathbf{J} + \frac{1}{c}\frac{\partial \mathbf{D}}{\partial t}
\end{aligned}
}
$$

Где:
- $\mathbf{E}$, $\mathbf{D}$ - напряженность и электрическая индукция;
- $\mathbf{H}$, $\mathbf{B}$ - напряженность магнитного поля и магнитная индукция;
- $\rho$ - объемная плотность заряда;
- $\mathbf{J}$ - плотность тока;
- $c$ - скорость света.

Для линейной изотропной среды:

$$ \mathbf{D}=\varepsilon\mathbf{E}, \qquad \mathbf{B}=\mu\mathbf{H}, \qquad \mathbf{J}=\sigma\mathbf{E}. $$

---

## 1. Закон Гаусса для электричества (СГС)

**Уравнение:**
$$ \nabla \cdot \mathbf{D} = 4\pi\rho $$

**Операторы в трёхмерном виде:**
- $\nabla = \left( \dfrac{\partial}{\partial x}, \dfrac{\partial}{\partial y}, \dfrac{\partial}{\partial z} \right)$ — векторный оператор частных производных.
- $\nabla \cdot \mathbf{D} = \dfrac{\partial D_x}{\partial x} + \dfrac{\partial D_y}{\partial y} + \dfrac{\partial D_z}{\partial z}$ — **дивергенция**.

**Физический смысл:**  
В системе СГС дивергенция электрической индукции равна $4\pi$, умноженным на плотность заряда. Это определяет поток поля через замкнутую поверхность.

---

## 2. Закон Гаусса для магнетизма

**Уравнение:**
$$ \nabla \cdot \mathbf{B} = 0 $$

**Физический смысл:**  
Отсутствие магнитных монополей. Линии магнитной индукции всегда замкнуты.

---

## 3. Закон электромагнитной индукции Фарадея (СГС)

**Уравнение:**
$$ \nabla \times \mathbf{E} = -\frac{1}{c}\frac{\partial \mathbf{B}}{\partial t} $$

**Операторы в трёхмерном виде:**
- $\nabla \times \mathbf{E}$ — **ротор**. Вычисляется как определитель:
  $$ \nabla \times \mathbf{E} =
  \begin{vmatrix}
  \mathbf{i} & \mathbf{j} & \mathbf{k} \\
  \dfrac{\partial}{\partial x} & \dfrac{\partial}{\partial y} & \dfrac{\partial}{\partial z} \\
  E_x & E_y & E_z
  \end{vmatrix} =
  \left( \frac{\partial E_z}{\partial y} - \frac{\partial E_y}{\partial z},\;
         \frac{\partial E_x}{\partial z} - \frac{\partial E_z}{\partial x},\;
         \frac{\partial E_y}{\partial x} - \frac{\partial E_x}{\partial y} \right)
  $$

**Физический смысл:**  
Изменение магнитной индукции порождает вихревое электрическое поле. Коэффициент $1/c$ связывает единицы измерения электрических и магнитных величин в СГС.

---

## 4. Закон Ампера — Максвелла (СГС)

**Уравнение:**
$$ \nabla \times \mathbf{H} = \frac{4\pi}{c}\mathbf{J} + \frac{1}{c}\frac{\partial \mathbf{D}}{\partial t} $$

**Физический смысл:**  
Магнитное поле создаётся токами проводимости (умноженными на $4\pi/c$) и токами смещения. В СГС ток смещения определяется как $\frac{1}{c}\frac{\partial \mathbf{D}}{\partial t}$.

---

### Параметры среды

*   **$\varepsilon$ (диэлектрическая проницаемость)**: Этот параметр характеризует способность среды поляризоваться под действием электрического поля. Он показывает, во сколько раз электрическое поле в среде ослабляется по сравнению с полем в вакууме. Чем больше $\varepsilon$, тем сильнее поляризуется среда и тем эффективнее она противодействует внешнему электрическому полю. Для вакуума $\varepsilon=1$ (в СГС). В общем случае, $\varepsilon$ может быть комплексной величиной, отражая потери энергии в среде (например, из-за поглощения).

*   **$\mu$ (магнитная проницаемость)**: Этот параметр описывает, как среда реагирует на магнитное поле. Он показывает, во сколько раз магнитное поле в среде усиливается (или ослабляется) по сравнению с полем в вакууме. Для большинства диэлектриков и немагнитных материалов $\mu \approx 1$ (в СГС). Для ферромагнитных материалов $\mu$ может быть значительно больше единицы. В общем случае, $\mu$ также может быть комплексной величиной.

*   **$\sigma$ (удельная электропроводность)**: Этот коэффициент определяет способность материала проводить электрический ток. Чем больше $\sigma$, тем лучше материал проводит ток. При наличии проводимости $\mathbf{J}=\sigma\mathbf{E}$, что приводит к диссипации энергии (нагреву) в среде. В идеальных диэлектриках $\sigma=0$, а в идеальных проводниках $\sigma\to\infty$.

## 2) Граничные условия на интерфейсе двух сред [1]


> **1.1.3. Граничные условия на поверхностях раздела (Борн, Вольф)**
>
> Согласно выводу из интегральных уравнений Максвелла, на поверхности раздела двух сред выполняются следующие условия:
>
> 1. **Нормальная компонента вектора магнитной индукции** (15):
> $$\mathbf{n}_{12} \cdot (\mathbf{B}^{(2)} - \mathbf{B}^{(1)}) = 0$$
>
> 2. **Нормальная компонента вектора электрического смещения** (19):
> $$\mathbf{n}_{12} \cdot (\mathbf{D}^{(2)} - \mathbf{D}^{(1)}) = 4\pi\hat{\rho}$$
> *(где $\hat{\rho}$ — поверхностная плотность заряда)*
>
> 3. **Тангенциальная компонента электрического вектора** (23):
> $$\mathbf{n}_{12} \times (\mathbf{E}^{(2)} - \mathbf{E}^{(1)}) = 0$$
>
> 4. **Тангенциальная компонента магнитного вектора** (25):
> $$\mathbf{n}_{12} \times (\mathbf{H}^{(2)} - \mathbf{H}^{(1)}) = \frac{4\pi}{c} \mathbf{j}$$
> *(где $\mathbf{j}$ — поверхностная плотность тока)*

Физический смысл:
- нормальная компонента $\mathbf{D}$ испытывает скачок из-за поверхностного заряда;
- нормальная компонента $\mathbf{B}$ непрерывна (магнитных монополей нет);
- тангенциальная компонента $\mathbf{E}$ непрерывна;
- тангенциальная компонента $\mathbf{H}$ может иметь скачок из-за поверхностного тока.

Эти условия соединяют решения Максвелла в соседних слоях тонкопленочной системы.

## 2.1) Определение плоской волны (Борн, Вольф)

Согласно подразделу 1.4.1 [1], гармоническая плоская волна описывается следующим образом:

Каждая из декартовых компонент векторов $\mathbf{E}$ и $\mathbf{H}$ имеет вид:

$$ a \cos(\tau + \delta) = \text{Re} \{ a \exp [-i(\tau + \delta)] \} \quad (a > 0) \qquad (10) $$

Здесь $\tau$ обозначает переменную часть фазового множителя:

$$ \tau = \omega \left( t - \frac{\mathbf{n} \cdot \mathbf{r}}{v} \right) \qquad (11) $$

Где:
- $\omega$ — круговая частота;
- $\mathbf{n}$ — единичный вектор в направлении распространения;
- $v$ — фазовая скорость в среде.


## 3) Как работает ротор $\nabla\times\mathbf{E}$ [1]

Для поля $\mathbf{E}=(E_x,E_y,E_z)$ компонентная запись ротора:

$$
\nabla\times\mathbf{E}=
\left(
\frac{\partial E_z}{\partial y}-\frac{\partial E_y}{\partial z},
\frac{\partial E_x}{\partial z}-\frac{\partial E_z}{\partial x},
\frac{\partial E_y}{\partial x}-\frac{\partial E_x}{\partial y}
\right).
$$

Рассмотрим $z$-компоненту:

$$
(\nabla\times\mathbf{E})_z=\frac{\partial E_y}{\partial x}-\frac{\partial E_x}{\partial y}.
$$

Пример:

$$
\mathbf{E}=(-y,\,x,\,0).
$$

Шаги вычисления:
1. $\dfrac{\partial E_y}{\partial x}=\dfrac{\partial (x)}{\partial x}=1$.
2. $\dfrac{\partial E_x}{\partial y}=\dfrac{\partial (-y)}{\partial y}=-1$.
3. $$
(\nabla\times\mathbf{E})_z=1-(-1)=2.
$$

Значит локальная "завихренность" по оси $z$ положительна. По закону Фарадея это связано с изменением магнитного поля:

$$
\nabla\times\mathbf{E}=-\frac{1}{c}\frac{\partial\mathbf{B}}{\partial t}.
$$

То есть ненулевой ротор электрического поля возникает, когда $\mathbf{B}$ меняется во времени.

## Как работает оператор Набла $\nabla$

### Пример 1: Градиент скалярного поля
Пусть задано скалярное поле потенциала $\phi(x, y, z) = x^2 + 2yz$.
Найдём вектор градиента $\nabla\phi$:
1.  Производная по $x$: $\frac{\partial}{\partial x}(x^2 + 2yz) = 2x$.
2.  Производная по $y$: $\frac{\partial}{\partial y}(x^2 + 2yz) = 2z$.
3.  Производная по $z$: $\frac{\partial}{\partial z}(x^2 + 2yz) = 2y$.

**Итог:** $\nabla\phi = (2x, 2z, 2y)$. Это вектор, указывающий направление роста потенциала.

### Пример 2: Дивергенция векторного поля
Возьмём векторное поле $\mathbf{A} = (x^2, y^2, z^2)$. Найдём его дивергенцию $\nabla \cdot \mathbf{A}$:
1.  Дифференцируем $A_x$ по $x$: $\frac{\partial}{\partial x}(x^2) = 2x$.
2.  Дифференцируем $A_y$ по $y$: $\frac{\partial}{\partial y}(y^2) = 2y$.
3.  Дифференцируем $A_z$ по $z$: $\frac{\partial}{\partial z}(z^2) = 2z$.
4.  Складываем результаты: $\nabla \cdot \mathbf{A} = 2x + 2y + 2z$.

**Итог:** Получено скалярное выражение, которое описывает интенсивность «источника» поля в каждой точке.

## 4) Свойства оператора Набла для плоской волны и уравнения Максвелла [1]

Рассмотрим гармоническую плоскую волну:

Из определения Борна и Вольфа (п. 2.1):
$$\tau = \omega \left( t - \frac{\mathbf{n} \cdot \mathbf{r}}{v} \right)$$

Раскроем скобки и введем волновое число $k = \frac{\omega}{v}$:
$$\tau = \omega t - \frac{\omega}{v}(\mathbf{n} \cdot \mathbf{r}) = \omega t - k(\mathbf{n} \cdot \mathbf{r})$$

Определим волновой вектор как $\mathbf{k} = k\mathbf{n}$. Тогда:
$$\tau = \omega t - \mathbf{k} \cdot \mathbf{r}$$

**Определения величин:**
*   $k = \frac{\omega}{v} = \frac{2\pi}{\lambda}$ — **волновое число**, определяющее пространственную периодичность волны (количество циклов на единицу длины, умноженное на $2\pi$).
*   $\mathbf{r} = (x, y, z)$ — **радиус-вектор**, определяющий координаты точки в пространстве.
*   $v$ — **фазовая скорость** распространения волны в данной среде.
*   $\omega = 2\pi f$ — **круговая частота**, характеризующая скорость изменения фазы во времени.

Подставим это в комплексную форму (10):
$$\mathbf{E} = \text{Re} \{ \mathbf{E}_0 e^{-i(\tau + \delta)} \} = \text{Re} \{ \mathbf{E}_0 e^{-i(\omega t - \mathbf{k} \cdot \mathbf{r} + \delta)} \}$$

Объединяя начальную фазу $\delta$ с амплитудой $\mathbf{E}_0$ и переставляя знаки в экспоненте (физический смысл не меняется при выборе знака фазы), получаем стандартный вид, используемый в расчетах:

$$\mathbf{E}(\mathbf{r},t)=\mathbf{E}_0\,e^{i(\mathbf{k}\cdot\mathbf{r}-\omega t)}, \qquad \mathbf{k}\cdot\mathbf{r}=k_xx+k_yy+k_zz.$$


### 4.1 Почему $\nabla\to i\mathbf{k}$

Покажем на примере компоненты $E_x$:
$$E_x=E_{0x}e^{i(k_xx+k_yy+k_zz-\omega t)}.$$

Берем частную производную по $x$:
$$\frac{\partial E_x}{\partial x} = E_{0x}\,\frac{\partial}{\partial x}\left(e^{i(k_xx+k_yy+k_zz-\omega t)}\right) = E_{0x}\,(ik_x)e^{i(\cdots)} = ik_xE_x.$$

Аналогично для других осей:
$$\frac{\partial E_x}{\partial y}=ik_yE_x, \qquad \frac{\partial E_x}{\partial z}=ik_zE_x.$$

Следовательно, действие оператора градиента $\nabla$ на поле плоской волны эквивалентно умножению на вектор $i\mathbf{k}$:
$$\nabla = (\partial_x, \partial_y, \partial_z) \Longrightarrow i(k_x, k_y, k_z) = i\mathbf{k}.$$

### 4.2 Почему $\partial_t\to -i\omega$

Берем производную по времени:
$$\frac{\partial E_x}{\partial t} = E_{0x}\,\frac{\partial}{\partial t}\left(e^{i(\cdots-\omega t)}\right) = E_{0x}\,(-i\omega)e^{i(\cdots)} = -i\omega E_x.$$

Таким образом: $\frac{\partial}{\partial t} \Longrightarrow -i\omega$.

---

### 4.3 Подстановка в уравнения Максвелла для вакуума

В вакууме (без источников) уравнения Максвелла имеют вид:
$$\nabla\cdot\mathbf{E}=0, \quad \nabla\cdot\mathbf{B}=0, \quad \nabla\times\mathbf{E}=-\frac{1}{c}\frac{\partial\mathbf{B}}{\partial t}, \quad \nabla\times\mathbf{B}=\frac{1}{c}\frac{\partial\mathbf{E}}{\partial t}.$$

Применяя замены $\nabla\to i\mathbf{k}$ и $\partial_t\to -i\omega$, получаем алгебраические уравнения:

1.  **Поперечность поля E:**
    $$i\mathbf{k}\cdot\mathbf{E}_0=0 \Rightarrow \mathbf{k}\perp\mathbf{E}_0.$$
2.  **Поперечность поля B:**
    $$i\mathbf{k}\cdot\mathbf{B}_0=0 \Rightarrow \mathbf{k}\perp\mathbf{B}_0.$$
3.  **Связь E и B (закон Фарадея):**
    $$i\mathbf{k}\times\mathbf{E}_0 = -\frac{1}{c}(-i\omega)\mathbf{B}_0 = i\frac{\omega}{c}\mathbf{B}_0 \Rightarrow \mathbf{B}_0 = \frac{c}{\omega}(\mathbf{k}\times\mathbf{E}_0).$$

### 4.4 Итоги для вакуума
Так как в вакууме волновое число $k = |\mathbf{k}| = \omega/c$, из векторного произведения следует:
*   $\mathbf{B}_0 = \frac{1}{k}(\mathbf{k}\times\mathbf{E}_0)$;
*   Векторы $\mathbf{E}$, $\mathbf{B}$ и $\mathbf{k}$ взаимно перпендикулярны;
*   В системе СГС амплитуды численного равны: $|\mathbf{E}_0| = |\mathbf{B}_0|$.
